# 02 — Model Development and Evaluation
This notebook runs the reproducible training pipeline in `src/train.py`. It compares classical baselines with the Keras neural network.

```python
from pathlib import Path
import subprocess, sys

repo_root = Path.cwd()
if not (repo_root / 'src' / 'train.py').exists():
    repo_root = repo_root.parent
assert (repo_root / 'src' / 'train.py').exists(), 'Open this notebook from the repository or its notebooks folder.'
result = subprocess.run([sys.executable, '-m', 'src.train', '--data', 'Data/loan_data.csv', '--output', 'artifacts'], cwd=repo_root, check=True, text=True, capture_output=True)
print(result.stdout)
print(result.stderr)
```

## Review test-set metrics
The held-out test set is evaluated after threshold selection on validation data. Compare recall and precision for class 1, not accuracy alone.

```python
import pandas as pd
import json
from IPython.display import display, Image
artifact_dir = repo_root / 'artifacts'
display(pd.read_csv(artifact_dir / 'model_comparison.csv'))
metrics = json.loads((artifact_dir / 'test_metrics.json').read_text())
for item in metrics:
    print('\\n', item['model'], '| threshold:', round(item['threshold'], 3))
    print(pd.DataFrame(item['confusion_matrix'], index=['actual_0','actual_1'], columns=['pred_0','pred_1']))
display(Image(filename=str(artifact_dir / 'roc_curve.png')))
display(Image(filename=str(artifact_dir / 'precision_recall_curve.png')))
```
